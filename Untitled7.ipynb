```python
import pandas as pd
```

```python
data = pd.read_csv('/content/xy_data (1).csv')
```

```python
print(data.head())
print("Number of rows:", len(data))
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(data['x'], data['y'], s=5)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Given XY Data')
plt.grid(True)
plt.show()
```

```python
import numpy as np
from scipy.optimize import least_squares

# Get x and y values from the CSV
x = data['x'].to_numpy()
y = data['y'].to_numpy()


# Function that calculates the error for a given
# theta, M and X
def residuals(params):

    theta, M, X = params

    # theta is in radians for Python's sin/cos
    cos_theta = np.cos(theta)
    sin_theta = np.sin(theta)

    # Calculate t from the observed x,y points
    t = (x - X) * cos_theta + (y - 42) * sin_theta

    # Calculate the oscillation component from x,y
    A = -(x - X) * sin_theta + (y - 42) * cos_theta

    # Expected oscillation component
    A_expected = np.exp(M * np.abs(t)) * np.sin(0.3 * t)

    # Difference between observed and expected
    return A - A_expected


# Initial guess
initial_guess = [
    np.deg2rad(25),   # theta = 25 degrees
    0.0,              # M
    50.0              # X
]


# Bounds given in the assignment
lower_bounds = [
    np.deg2rad(0),    # theta > 0 degrees
    -0.05,            # M
    0.0               # X
]

upper_bounds = [
    np.deg2rad(50),   # theta < 50 degrees
    0.05,             # M
    100.0             # X
]


# Perform numerical optimization
result = least_squares(
    residuals,
    initial_guess,
    bounds=(lower_bounds, upper_bounds)
)


# Extract the results
theta = result.x[0]
M = result.x[1]
X = result.x[2]


print("Estimated parameters:")
print("Theta =", np.rad2deg(theta), "degrees")
print("M     =", M)
print("X     =", X)

print("\nOptimization error:")
print(np.sum(np.abs(result.fun)))
```

```python
# Use the clean final parameter values
theta_final = np.deg2rad(30)
M_final = 0.03
X_final = 55

# Calculate t for every given data point
t_values = (
    (x - X_final) * np.cos(theta_final)
    + (y - 42) * np.sin(theta_final)
)

# Generate predicted x and y using the original equations
x_pred = (
    t_values * np.cos(theta_final)
    - np.exp(M_final * np.abs(t_values))
    * np.sin(0.3 * t_values)
    * np.sin(theta_final)
    + X_final
)

y_pred = (
    42
    + t_values * np.sin(theta_final)
    + np.exp(M_final * np.abs(t_values))
    * np.sin(0.3 * t_values)
    * np.cos(theta_final)
)

# Calculate absolute errors
x_error = np.abs(x - x_pred)
y_error = np.abs(y - y_pred)

# L1 error for every point
l1_error = x_error + y_error

print("Verification Results")
print("--------------------")
print("Maximum X error:", np.max(x_error))
print("Maximum Y error:", np.max(y_error))
print("Mean L1 error:", np.mean(l1_error))
print("Total L1 error:", np.sum(l1_error))

print("\nRange of calculated t:")
print("Minimum t:", np.min(t_values))
print("Maximum t:", np.max(t_values))
```

```python
plt.figure(figsize=(10, 6))

# Original CSV points
plt.scatter(
    x, y,
    s=8,
    label='Given CSV Data'
)

# Sort by t so the predicted curve is drawn correctly
order = np.argsort(t_values)

plt.plot(
    x_pred[order],
    y_pred[order],
    color='red',
    linewidth=2,
    label='Predicted Curve'
)

plt.xlabel('x')
plt.ylabel('y')
plt.title('Given Data vs Predicted Parametric Curve')
plt.legend()
plt.grid(True)
plt.show()
```
